```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("Text-to-PDF-yVx.pdf")
docs = loader.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=20, chunk_overlap=10)
final_docs = text_splitter.split_documents(docs)
final_docs
```

```python
print(final_docs[0])
print(final_docs[1])
```

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("speech.txt")
text_documents = loader.load()
text_documents
# type(text_documents)
```

```python
from langchain_text_splitters import CharacterTextSplitter
text_splitter = CharacterTextSplitter(separator="\n\n",chunk_size=30, chunk_overlap=10)
texts = text_splitter.split_documents(text_documents)
texts[0]
```

```python
with open("speech.txt", "r") as file:
    content = file.read()
content

text_splitter=RecursiveCharacterTextSplitter(chunk_size=30, chunk_overlap=10)
texts=text_splitter.create_documents([content])
texts[0]
texts[1]
```
