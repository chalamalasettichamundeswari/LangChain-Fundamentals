FAISS

Faiss stands for "Facebook AI Similarity Search." It is a library developed by Facebook AI Research for efficient similarity search and clustering of dense vectors, commonly used for fast nearest neighbor search in machine learning and AI applications.

Faiss is a tool for quickly finding similar items in large collections of data. It is mainly used to search for vectors that are close to each other, which is useful for recommendations, search engines, and machine learning tasks.

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain.text_splitter import CharacterTextSplitter

loader = TextLoader("speech.txt")
documents = loader.load()
text_splitter = CharacterTextSplitter(chunk_size=10, chunk_overlap=5)
docs = text_splitter.split_documents(documents)

embeddings = OllamaEmbeddings(model="gemma:2b")
db = FAISS.from_documents(docs, embeddings)
# db

##quering
query="from startups to global"
docsSearch = db.similarity_search(query)
docsSearch

##retrieving- convert vectorstore into a Retriever class. This allows us to easily use it in other LangChain methods, which largely work with retrievers.
retriever = db.as_retriever()
docsRetrieve = retriever.invoke(query)
docsRetrieve[0].query
```
