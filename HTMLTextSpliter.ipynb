```python
from langchain_text_splitters import HTMLHeaderTextSplitter

html_string="""
<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="utf-8" />
  <meta name="viewport" content="width=device-width, initial-scale=1" />
  <title>Sample HTML — No CSS</title>
  <meta name="description" content="A minimal, semantic HTML page with common elements and no CSS." />
</head>
<body>
  <!-- Header & Navigation -->
  <header>
    <h1>My Sample HTML Page</h1>
    <h2>Menu</h2>
    <h3>Navigation Links</h3>
  </header>
</body>
</html>
"""

headers_to_split_on = [("h1","heading"),("h2","subheading")]
html_splitter = HTMLHeaderTextSplitter(headers_to_split_on=headers_to_split_on)
html_header_splitter = html_splitter.split_text(html_string)
html_header_splitter
```

```python
url="http://linkedin.com/"
headers_to_split_on = [("h1","heading")]
html_splitter = HTMLHeaderTextSplitter(headers_to_split_on=headers_to_split_on)
html_header_splitter = html_splitter.split_text_from_url(url)
html_header_splitter
```
