```python
##building a sample vectordb
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import OllamaEmbeddings
from langchain.text_splitter import RecursiveCharacterTextSplitter

loader = TextLoader('speech.txt')
data = loader.load()
#data

text_splitter = RecursiveCharacterTextSplitter(chunk_size=30, chunk_overlap=5)
splits = text_splitter.split_documents(data)
#splits

embedding = OllamaEmbeddings(model="gemma:2b")
vectordb = Chroma.from_documents(splits, embedding)
vectordb

##query it
query="from startups to global"
docsSearch = vectordb.similarity_search(query)

##saving to the disk
vectordb = Chroma.from_documents(splits, embedding, persist_directory="./chroma_db")

##load from disk
vectordb1 = Chroma(persist_directory="./chroma_db", embedding_function=embedding)
# vectordb2 = vectordb1.similarity_search(query)
vectordb2 = vectordb1.similarity_search_with_score(query)
#vectordb2[0].page_content
vectordb2

##Retriever option
retriever = vectordb1.as_retriever()
retriever.invoke(query)[0].page_content
```
